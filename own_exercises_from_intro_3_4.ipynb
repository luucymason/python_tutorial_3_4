Q6. Boolean conditions and lambda
	1. Create a High_Value column that is True when revenue exceeds £500 and False otherwise.
	2. Use .apply() and a lambda expression to create a Feedback column:
		○ Rating above 3 = "Positive"
		○ Rating of 3 or below = "Needs Review"
Display the relevant columns to check your results.

```python
import pandas as pd
import numpy as np
orders = pd.DataFrame({
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006],
    "Customer": ["Alice", "Ben", "Chloe", "David", "Ella", "Finn"],
    "Product": ["Laptop", "Mouse", "Laptop", "Keyboard", "Mouse", "Monitor"],
    "Price": [899, 25, 899, 75, 25, 220],
    "Quantity": [1, 2, 1, 1, 4, 2],
    "Region": ["North", "South", "North", "West", "East", "South"],
    "Rating": [5, 4, 5, 3, 4, 5]
})
customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 4, 5, 6],
    "Country": [" UK", "usa", "UK ", "Germany", "USA ", "germany"],
    "Age": ["22", "35", "28", "41", "35", "19"],
    "Spend": [120.5, 300.0, np.nan, 450.0, 300.0, np.nan],
    "Membership": ["Gold", "Silver", "Gold", "Bronze", "Silver", "Gold"]
})

orders["Revenue"] = orders["Price"]*orders["Quantity"]

#High_Value column when revenue is true when bigger than 500 and false if else

orders["High_Value"] = orders["Revenue"].apply(lambda x: x > 500)
print(orders)

#make feedback column with .apply and lambda where rating above 3 is positive and below 3 is needs review

orders["Feedback"] = orders["Rating"].apply(lambda x: "Positive" if x > 3 else "Needs review")
print(orders)
```

Q10. Standardising text and data types

	2. Convert country names to uppercase.
	3. Convert Age from text into integers using .astype(int).
	4. Calculate the average age.

```python
import pandas as pd
import numpy as np
orders = pd.DataFrame({
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006],
    "Customer": ["Alice", "Ben", "Chloe", "David", "Ella", "Finn"],
    "Product": ["Laptop", "Mouse", "Laptop", "Keyboard", "Mouse", "Monitor"],
    "Price": [899, 25, 899, 75, 25, 220],
    "Quantity": [1, 2, 1, 1, 4, 2],
    "Region": ["North", "South", "North", "West", "East", "South"],
    "Rating": [5, 4, 5, 3, 4, 5]
})
customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 4, 5, 6],
    "Country": [" UK", "usa", "UK ", "Germany", "USA ", "germany"],
    "Age": ["22", "35", "28", "41", "35", "19"],
    "Spend": [120.5, 300.0, np.nan, 450.0, 300.0, np.nan],
    "Membership": ["Gold", "Silver", "Gold", "Bronze", "Silver", "Gold"]
})
orders["Revenue"] = orders["Price"]*orders["Quantity"]

#remove whitespace from country

customers["Country"] = customers["Country"].str.strip()

#convert to uppercase

customers["Country"] = customers["Country"].str.upper()

#convert age from text into integers using .astype(int)

customers["Age"] = customers["Age"].astype(int)
print(customers)

#calculate average age, chose to look at mean

customers["Age"].mean()
```

Q11. Duplicates and column names
	1. Create a duplicate row in a copy of customers.
	2. Count duplicates using .duplicated().sum().
	3. Remove duplicates and reset the index.
	4. Remove spaces from column names and convert them to lowercase.
Rename customer_id to id using .rename(columns={...}).

```python
import pandas as pd
import numpy as np
orders = pd.DataFrame({
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006],
    "Customer": ["Alice", "Ben", "Chloe", "David", "Ella", "Finn"],
    "Product": ["Laptop", "Mouse", "Laptop", "Keyboard", "Mouse", "Monitor"],
    "Price": [899, 25, 899, 75, 25, 220],
    "Quantity": [1, 2, 1, 1, 4, 2],
    "Region": ["North", "South", "North", "West", "East", "South"],
    "Rating": [5, 4, 5, 3, 4, 5]
})
customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 4, 5, 6],
    "Country": [" UK", "usa", "UK ", "Germany", "USA ", "germany"],
    "Age": ["22", "35", "28", "41", "35", "19"],
    "Spend": [120.5, 300.0, np.nan, 450.0, 300.0, np.nan],
    "Membership": ["Gold", "Silver", "Gold", "Bronze", "Silver", "Gold"]
})
orders["Revenue"] = orders["Price"]*orders["Quantity"]

#make duplicate row in a copy of customers (chose row 1)

customers_copy = customers.copy()
customers_copy = pd.concat(
    [customers_copy, customers_copy.iloc[[0]]],
    ignore_index = True
)


#.copy() makes a copy, iloc[[0]] selects first row as a dataframe
#pd.concat adds the duplicated row at the bottom
#nidex true resets the row index so it runs consecutively

#count duplicates using .duplicated().sum()
customers_copy.duplicated().sum()

#remove duplicates and reset index

customers_copy.drop_duplicates().reset_index(drop = True)

#remove spaces from column names and convert to lowercase
#Rename customer_id to id using .rename(columns={...})

customers.columns = customers.columns.str.strip()
customers.rename(columns={"Customer_ID": "ID"}, inplace = True)
print(customers)
```
